```python
from sklearn.model_selection import train_test_split

# Target
y_hf = heart_failure["DEATH_EVENT"]

# Remove target + follow-up time (to avoid temporal leakage)
X_hf = heart_failure.drop(columns=["DEATH_EVENT", "time"])

# 80/20 stratified split
X_train_hf, X_test_hf, y_train_hf, y_test_hf = train_test_split(
    X_hf,
    y_hf,
    test_size=0.20,
    stratify=y_hf,
    random_state=42
)

print("Training:", X_train_hf.shape)
print("Testing:", X_test_hf.shape)
print("Train target:", y_train_hf.value_counts().to_dict())
print("Test target:", y_test_hf.value_counts().to_dict())
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from xgboost import XGBClassifier

from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score, average_precision_score,
    confusion_matrix
)

models = {
    "Logistic Regression": Pipeline([
        ("scaler", StandardScaler()),
        ("model", LogisticRegression(max_iter=2000, random_state=42))
    ]),

    "SVM": Pipeline([
        ("scaler", StandardScaler()),
        ("model", SVC(probability=True, random_state=42))
    ]),

    "Decision Tree": DecisionTreeClassifier(
        random_state=42, class_weight="balanced"
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=500,
        random_state=42,
        class_weight="balanced"
    ),

    "XGBoost": XGBClassifier(
        n_estimators=300,
        max_depth=3,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        eval_metric="logloss",
        random_state=42
    ),

    "KNN": Pipeline([
        ("scaler", StandardScaler()),
        ("model", KNeighborsClassifier(n_neighbors=7))
    ])
}

results = []

for name, model in models.items():

    model.fit(X_train_hf, y_train_hf)

    pred = model.predict(X_test_hf)
    prob = model.predict_proba(X_test_hf)[:, 1]

    tn, fp, fn, tp = confusion_matrix(
        y_test_hf, pred
    ).ravel()

    specificity = tn / (tn + fp)

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test_hf, pred),
        "Precision": precision_score(y_test_hf, pred, zero_division=0),
        "Recall": recall_score(y_test_hf, pred, zero_division=0),
        "Specificity": specificity,
        "F1": f1_score(y_test_hf, pred, zero_division=0),
        "ROC_AUC": roc_auc_score(y_test_hf, prob),
        "PR_AUC": average_precision_score(y_test_hf, prob)
    })

hf_results = pd.DataFrame(results).round(4)

display(hf_results)
```

```python
from sklearn.model_selection import RepeatedStratifiedKFold, cross_validate
from sklearn.metrics import make_scorer, recall_score

cv = RepeatedStratifiedKFold(
    n_splits=5,
    n_repeats=5,
    random_state=42
)

scoring = {
    "Accuracy": "accuracy",
    "Precision": "precision",
    "Recall": "recall",
    "F1": "f1",
    "ROC_AUC": "roc_auc",
    "PR_AUC": "average_precision"
}

cv_results = []

for name, model in models.items():

    scores = cross_validate(
        model,
        X_hf,
        y_hf,
        cv=cv,
        scoring=scoring,
        n_jobs=-1
    )

    cv_results.append({
        "Model": name,
        "Accuracy": f"{scores['test_Accuracy'].mean():.3f} ± {scores['test_Accuracy'].std():.3f}",
        "Precision": f"{scores['test_Precision'].mean():.3f} ± {scores['test_Precision'].std():.3f}",
        "Recall": f"{scores['test_Recall'].mean():.3f} ± {scores['test_Recall'].std():.3f}",
        "F1": f"{scores['test_F1'].mean():.3f} ± {scores['test_F1'].std():.3f}",
        "ROC_AUC": f"{scores['test_ROC_AUC'].mean():.3f} ± {scores['test_ROC_AUC'].std():.3f}",
        "PR_AUC": f"{scores['test_PR_AUC'].mean():.3f} ± {scores['test_PR_AUC'].std():.3f}"
    })

hf_cv_results = pd.DataFrame(cv_results)

display(hf_cv_results)
```

```python
# ============================================================
# FIGURE: ROC CURVES — HEART FAILURE DATASET
# Publication-ready version
# ============================================================

from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import roc_curve, auc
import matplotlib.pyplot as plt
import numpy as np

cv5 = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

plt.figure(figsize=(9, 7))

for name, model in models.items():

    y_prob = np.zeros(len(y_hf))

    for train_idx, val_idx in cv5.split(X_hf, y_hf):

        Xtr = X_hf.iloc[train_idx]
        Xval = X_hf.iloc[val_idx]
        ytr = y_hf.iloc[train_idx]

        model.fit(Xtr, ytr)

        y_prob[val_idx] = model.predict_proba(Xval)[:, 1]

    fpr, tpr, _ = roc_curve(y_hf, y_prob)
    roc_auc = auc(fpr, tpr)

    plt.plot(
        fpr,
        tpr,
        linewidth=2,
        label=f"{name} (AUC = {roc_auc:.3f})"
    )

# Reference line
plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    linewidth=1.5,
    label="Random classifier (AUC = 0.500)"
)

# ---- Publication-ready labels ----
plt.xlabel(
    "False Positive Rate (1 − Specificity)",
    fontsize=13
)

plt.ylabel(
    "True Positive Rate (Sensitivity)",
    fontsize=13
)

plt.title(
    "Receiver Operating Characteristic Curves for Heart Failure Mortality Prediction",
    fontsize=15,
    fontweight="bold",
    pad=15
)

plt.legend(
    loc="lower right",
    fontsize=10,
    frameon=True
)

plt.xlim(0, 1)
plt.ylim(0, 1.02)

plt.xticks(fontsize=11)
plt.yticks(fontsize=11)

plt.grid(
    True,
    linestyle=":",
    linewidth=0.7,
    alpha=0.6
)

plt.tight_layout()

# ---- Save publication-quality files ----
plt.savefig(
    "Figure_1_ROC_Heart_Failure.png",
    dpi=300,
    bbox_inches="tight"
)

plt.savefig(
    "Figure_1_ROC_Heart_Failure.pdf",
    bbox_inches="tight"
)

plt.show()
```

```python
# ============================================================
# FIGURE 4: CONFUSION MATRICES
# HEART FAILURE DATASET
# 5-FOLD STRATIFIED OUT-OF-FOLD PREDICTIONS
# Publication-ready version
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.base import clone
from sklearn.metrics import confusion_matrix


# ------------------------------------------------------------
# 1. Define the same 5-fold CV strategy used in Figures 1–3
# ------------------------------------------------------------

skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)


# ------------------------------------------------------------
# 2. Generate OUT-OF-FOLD predictions for all six models
# ------------------------------------------------------------

hf_oof_predictions = {}

for model_name, model in models.items():

    print(f"Generating OOF predictions for: {model_name}")

    model_clone = clone(model)

    y_pred_oof = cross_val_predict(
        model_clone,
        X_hf,
        y_hf,
        cv=skf,
        method="predict",
        n_jobs=-1
    )

    hf_oof_predictions[model_name] = y_pred_oof


# ------------------------------------------------------------
# 3. Store OOF predictions in a DataFrame
#    Useful for future analyses as well
# ------------------------------------------------------------

hf_oof_predictions_df = pd.DataFrame(
    hf_oof_predictions
)

hf_oof_predictions_df["Actual"] = np.asarray(y_hf)


# Optional: save OOF predictions
hf_oof_predictions_df.to_csv(
    "Heart_Failure_OOF_Predictions_5Fold.csv",
    index=False
)


# ------------------------------------------------------------
# 4. Model order
# ------------------------------------------------------------

model_order = [
    "Logistic Regression",
    "SVM",
    "Decision Tree",
    "Random Forest",
    "XGBoost",
    "KNN"
]


# ------------------------------------------------------------
# 5. Class labels
#    Heart Failure dataset:
#    0 = Survived / No Death Event
#    1 = Death Event
# ------------------------------------------------------------

class_labels = [0, 1]

class_names = [
    "No Death Event (0)",
    "Death Event (1)"
]


# ------------------------------------------------------------
# 6. Create 2 × 3 publication-ready figure
# ------------------------------------------------------------

fig, axes = plt.subplots(
    2,
    3,
    figsize=(15, 10)
)

axes = axes.flatten()


# ------------------------------------------------------------
# 7. Plot confusion matrix for each model
# ------------------------------------------------------------

for ax, model_name in zip(axes, model_order):

    y_pred = hf_oof_predictions[model_name]

    cm = confusion_matrix(
        y_hf,
        y_pred,
        labels=class_labels
    )

    # Display matrix
    im = ax.imshow(
        cm,
        interpolation="nearest",
        cmap="Blues"
    )

    # --------------------------------------------------------
    # Cell annotations: TN, FP, FN, TP
    # --------------------------------------------------------

    cell_labels = [
        ["TN", "FP"],
        ["FN", "TP"]
    ]

    for i in range(2):
        for j in range(2):

            ax.text(
                j,
                i,
                f"{cell_labels[i][j]}\n{cm[i, j]:,}",
                ha="center",
                va="center",
                fontsize=13,
                fontweight="bold"
            )

    # --------------------------------------------------------
    # Axis labels
    # --------------------------------------------------------

    ax.set_xticks([0, 1])
    ax.set_yticks([0, 1])

    ax.set_xticklabels(
        class_names,
        fontsize=10,
        rotation=15,
        ha="right"
    )

    ax.set_yticklabels(
        class_names,
        fontsize=10
    )

    ax.set_xlabel(
        "Predicted Class",
        fontsize=11,
        labelpad=8
    )

    ax.set_ylabel(
        "Actual Class",
        fontsize=11,
        labelpad=8
    )

    # --------------------------------------------------------
    # Subplot title
    # --------------------------------------------------------

    ax.set_title(
        model_name,
        fontsize=13,
        fontweight="bold",
        pad=12
    )

    # --------------------------------------------------------
    # Improve visual boundaries
    # --------------------------------------------------------

    ax.set_xticks(
        np.arange(-0.5, 2, 1),
        minor=True
    )

    ax.set_yticks(
        np.arange(-0.5, 2, 1),
        minor=True
    )

    ax.grid(
        which="minor",
        color="white",
        linestyle="-",
        linewidth=2
    )

    ax.tick_params(
        which="minor",
        bottom=False,
        left=False
    )


# ------------------------------------------------------------
# 8. Overall figure title
# ------------------------------------------------------------

fig.suptitle(
    "Confusion Matrices of Machine Learning Models for "
    "Heart Failure Mortality Prediction",
    fontsize=17,
    fontweight="bold",
    y=0.98
)


# ------------------------------------------------------------
# 9. Shared colorbar
# ------------------------------------------------------------

cbar = fig.colorbar(
    im,
    ax=axes,
    fraction=0.025,
    pad=0.02
)

cbar.set_label(
    "Number of Observations",
    fontsize=11
)

cbar.ax.tick_params(
    labelsize=10
)


# ------------------------------------------------------------
# 10. Figure note
# ------------------------------------------------------------

fig.text(
    0.5,
    0.015,
    "TN = True Negative; FP = False Positive; "
    "FN = False Negative; TP = True Positive. "
    "Matrices are based on 5-fold stratified out-of-fold predictions.",
    ha="center",
    fontsize=10
)


# ------------------------------------------------------------
# 11. Final layout
# ------------------------------------------------------------

plt.tight_layout(
    rect=[0, 0.055, 1, 0.94]
)


# ------------------------------------------------------------
# 12. Save publication-quality files
# ------------------------------------------------------------

plt.savefig(
    "Figure_4_Confusion_Matrices_Heart_Failure.png",
    dpi=300,
    bbox_inches="tight"
)

plt.savefig(
    "Figure_4_Confusion_Matrices_Heart_Failure.pdf",
    bbox_inches="tight"
)


# ------------------------------------------------------------
# 13. Display
# ------------------------------------------------------------

plt.show()
```

```python
# ============================================================
# FIGURE 5: CROSS-VALIDATED FEATURE IMPORTANCE
# HEART FAILURE DATASET
# DECISION TREE, RANDOM FOREST & XGBOOST
# Publication-ready version
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import StratifiedKFold
from sklearn.base import clone


# ------------------------------------------------------------
# 1. Five-fold stratified cross-validation
# ------------------------------------------------------------

skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)


# ------------------------------------------------------------
# 2. Models included
# ------------------------------------------------------------

importance_models = {
    "Decision Tree": models["Decision Tree"],
    "Random Forest": models["Random Forest"],
    "XGBoost": models["XGBoost"]
}


# ------------------------------------------------------------
# 3. Feature names
# ------------------------------------------------------------

feature_names = X_hf.columns.tolist()


# ------------------------------------------------------------
# 4. Calculate feature importance within each CV fold
# ------------------------------------------------------------

feature_importance_results = {}

for model_name, model in importance_models.items():

    fold_importances = []

    print(f"Calculating feature importance: {model_name}")

    for fold, (train_idx, test_idx) in enumerate(
        skf.split(X_hf, y_hf),
        start=1
    ):

        # Split data
        X_train = X_hf.iloc[train_idx]
        y_train = y_hf.iloc[train_idx]

        # Clone model so each fold is independent
        model_clone = clone(model)

        # Fit only on training fold
        model_clone.fit(
            X_train,
            y_train
        )

        # Native tree-based feature importance
        importances = model_clone.feature_importances_

        fold_importances.append(importances)

    # Convert to array
    fold_importances = np.array(
        fold_importances
    )

    # Mean importance across folds
    mean_importance = fold_importances.mean(axis=0)

    # SD across folds
    sd_importance = fold_importances.std(
        axis=0,
        ddof=1
    )

    # Store results
    feature_importance_results[model_name] = pd.DataFrame({
        "Feature": feature_names,
        "Mean_Importance": mean_importance,
        "SD_Importance": sd_importance
    })


# ------------------------------------------------------------
# 5. Display complete importance tables
# ------------------------------------------------------------

for model_name, importance_df in feature_importance_results.items():

    print("\n" + "=" * 70)
    print(model_name)
    print("=" * 70)

    display(
        importance_df.sort_values(
            "Mean_Importance",
            ascending=False
        ).reset_index(drop=True)
    )


# ------------------------------------------------------------
# 6. Select TOP 10 features for each model
# ------------------------------------------------------------

top_n = 10

top_features = {}

for model_name, importance_df in feature_importance_results.items():

    top_features[model_name] = (
        importance_df
        .sort_values(
            "Mean_Importance",
            ascending=False
        )
        .head(top_n)
        .sort_values(
            "Mean_Importance",
            ascending=True
        )
    )


# ------------------------------------------------------------
# 7. Create publication-ready 3-panel figure
# ------------------------------------------------------------

fig, axes = plt.subplots(
    nrows=1,
    ncols=3,
    figsize=(18, 8)
)


# ------------------------------------------------------------
# 8. Plot each model
# ------------------------------------------------------------

for ax, (model_name, df_top) in zip(
    axes,
    top_features.items()
):

    y_pos = np.arange(
        len(df_top)
    )

    bars = ax.barh(
        y_pos,
        df_top["Mean_Importance"],
        xerr=df_top["SD_Importance"],
        capsize=3,
        alpha=0.85
    )

    # Feature names
    ax.set_yticks(
        y_pos
    )

    ax.set_yticklabels(
        df_top["Feature"],
        fontsize=10
    )

    # X-axis
    ax.set_xlabel(
        "Mean Feature Importance",
        fontsize=11
    )

    # Title
    ax.set_title(
        model_name,
        fontsize=13,
        fontweight="bold",
        pad=12
    )

    # Value labels
    for bar, mean_value in zip(
        bars,
        df_top["Mean_Importance"]
    ):

        ax.text(
            bar.get_width() + 0.005,
            bar.get_y() + bar.get_height() / 2,
            f"{mean_value:.3f}",
            va="center",
            fontsize=9
        )

    # Grid
    ax.grid(
        axis="x",
        linestyle=":",
        linewidth=0.7,
        alpha=0.6
    )

    ax.tick_params(
        axis="x",
        labelsize=10
    )

    ax.tick_params(
        axis="y",
        labelsize=10
    )


# ------------------------------------------------------------
# 9. Overall figure title
# ------------------------------------------------------------

fig.suptitle(
    "Top 10 Predictive Features for Heart Failure Mortality "
    "Across Tree-Based Machine Learning Models",
    fontsize=16,
    fontweight="bold",
    y=1.02
)


# ------------------------------------------------------------
# 10. Figure note
# ------------------------------------------------------------

fig.text(
    0.5,
    -0.01,
    "Feature importance represents the mean importance across "
    "5-fold stratified cross-validation; error bars indicate "
    "the standard deviation across folds.",
    ha="center",
    fontsize=10
)


# ------------------------------------------------------------
# 11. Layout
# ------------------------------------------------------------

plt.tight_layout(
    rect=[0, 0.04, 1, 0.96]
)


# ------------------------------------------------------------
# 12. Save publication-quality files
# ------------------------------------------------------------

plt.savefig(
    "Figure_5_CrossValidated_Feature_Importance_Heart_Failure.png",
    dpi=300,
    bbox_inches="tight"
)

plt.savefig(
    "Figure_5_CrossValidated_Feature_Importance_Heart_Failure.pdf",
    bbox_inches="tight"
)


# ------------------------------------------------------------
# 13. Display
# ------------------------------------------------------------

plt.show()
```

```python
# ============================================================
# FIGURE 6: SHAP EXPLAINABILITY ANALYSIS
# HEART FAILURE DATASET
# 5-FOLD OUT-OF-FOLD SHAP VALUES
# DECISION TREE, RANDOM FOREST & XGBOOST
# Publication-ready version
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import shap

from sklearn.model_selection import StratifiedKFold
from sklearn.base import clone


# ------------------------------------------------------------
# 1. Five-fold stratified cross-validation
# ------------------------------------------------------------

skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)


# ------------------------------------------------------------
# 2. Models
# ------------------------------------------------------------

shap_models = {
    "Decision Tree": models["Decision Tree"],
    "Random Forest": models["Random Forest"],
    "XGBoost": models["XGBoost"]
}


# ------------------------------------------------------------
# 3. Feature names
# ------------------------------------------------------------

feature_names = X_hf.columns.tolist()


# ------------------------------------------------------------
# 4. Store OOF SHAP values
# ------------------------------------------------------------

oof_shap_values = {}

for model_name, model in shap_models.items():

    print(f"\nCalculating OOF SHAP values: {model_name}")

    shap_values_all = np.zeros(
        (len(X_hf), len(feature_names))
    )

    for fold, (train_idx, test_idx) in enumerate(
        skf.split(X_hf, y_hf),
        start=1
    ):

        print(f"  Fold {fold}/5")

        # ----------------------------------------------------
        # Training and test fold
        # ----------------------------------------------------

        X_train = X_hf.iloc[train_idx]
        y_train = y_hf.iloc[train_idx]

        X_test = X_hf.iloc[test_idx]

        # ----------------------------------------------------
        # Clone and fit model only on training fold
        # ----------------------------------------------------

        model_clone = clone(model)

        model_clone.fit(
            X_train,
            y_train
        )

        # ----------------------------------------------------
        # Tree SHAP
        # ----------------------------------------------------

        explainer = shap.TreeExplainer(
            model_clone
        )

        fold_shap = explainer.shap_values(
            X_test
        )

        # ----------------------------------------------------
        # Handle different SHAP output formats
        # ----------------------------------------------------

        if isinstance(fold_shap, list):

            # Binary classification
            fold_shap = fold_shap[1]

        elif len(np.asarray(fold_shap).shape) == 3:

            # Possible multiclass/binary 3D output
            fold_shap = fold_shap[:, :, 1]

        fold_shap = np.asarray(
            fold_shap
        )

        # ----------------------------------------------------
        # Store SHAP values in original observation positions
        # ----------------------------------------------------

        shap_values_all[test_idx, :] = fold_shap


    # Save
    oof_shap_values[model_name] = shap_values_all


print("\nOOF SHAP analysis completed successfully.")


# ------------------------------------------------------------
# 5. Calculate mean absolute SHAP importance
# ------------------------------------------------------------

shap_importance = {}

for model_name, shap_values in oof_shap_values.items():

    mean_abs_shap = np.mean(
        np.abs(shap_values),
        axis=0
    )

    importance_df = pd.DataFrame({
        "Feature": feature_names,
        "Mean_Absolute_SHAP": mean_abs_shap
    })

    importance_df = importance_df.sort_values(
        "Mean_Absolute_SHAP",
        ascending=False
    ).reset_index(drop=True)

    shap_importance[model_name] = importance_df


# ------------------------------------------------------------
# 6. Display SHAP importance tables
# ------------------------------------------------------------

for model_name, importance_df in shap_importance.items():

    print("\n" + "=" * 70)
    print(model_name)
    print("=" * 70)

    display(
        importance_df.head(10)
    )


# ------------------------------------------------------------
# 7. Select common top features
# ------------------------------------------------------------

# Take top 10 features from each model
top_features_by_model = {}

for model_name, importance_df in shap_importance.items():

    top_features_by_model[model_name] = (
        importance_df
        .head(10)["Feature"]
        .tolist()
    )


# ------------------------------------------------------------
# 8. Create publication-ready SHAP summary figure
# ------------------------------------------------------------

fig, axes = plt.subplots(
    1,
    3,
    figsize=(19, 9)
)


# ------------------------------------------------------------
# 9. Plot SHAP values manually
#    This allows all three models to be displayed together
# ------------------------------------------------------------

for ax, (model_name, shap_values) in zip(
    axes,
    oof_shap_values.items()
):

    # Top 10 features
    top_features = top_features_by_model[model_name]

    # Feature indices
    feature_indices = [
        feature_names.index(feature)
        for feature in top_features
    ]

    # Sort according to SHAP importance
    top_features = (
        shap_importance[model_name]
        .head(10)
        .sort_values(
            "Mean_Absolute_SHAP",
            ascending=True
        )["Feature"]
        .tolist()
    )

    # --------------------------------------------------------
    # Plot each feature
    # --------------------------------------------------------

    for y_position, feature in enumerate(
        top_features
    ):

        feature_index = feature_names.index(
            feature
        )

        shap_feature_values = shap_values[
            :,
            feature_index
        ]

        # Random vertical jitter
        rng = np.random.default_rng(
            42 + y_position
        )

        jitter = (
            rng.random(
                len(shap_feature_values)
            ) - 0.5
        ) * 0.65

        y = (
            np.full(
                len(shap_feature_values),
                y_position
            )
            + jitter
        )

        # Feature values for color
        feature_values = X_hf[
            feature
        ].values

        # Normalize feature values
        value_min = np.nanmin(
            feature_values
        )

        value_max = np.nanmax(
            feature_values
        )

        if value_max > value_min:

            normalized_values = (
                feature_values - value_min
            ) / (
                value_max - value_min
            )

        else:

            normalized_values = np.zeros(
                len(feature_values)
            )

        # Scatter
        ax.scatter(
            shap_feature_values,
            y,
            c=normalized_values,
            cmap="coolwarm",
            s=12,
            alpha=0.55,
            edgecolors="none"
        )

    # --------------------------------------------------------
    # Y-axis
    # --------------------------------------------------------

    ax.set_yticks(
        np.arange(
            len(top_features)
        )
    )

    ax.set_yticklabels(
        top_features,
        fontsize=9
    )

    # --------------------------------------------------------
    # X-axis
    # --------------------------------------------------------

    ax.set_xlabel(
        "SHAP Value",
        fontsize=11
    )

    # --------------------------------------------------------
    # Vertical zero line
    # --------------------------------------------------------

    ax.axvline(
        0,
        linestyle="--",
        linewidth=0.8,
        alpha=0.7
    )

    # --------------------------------------------------------
    # Title
    # --------------------------------------------------------

    ax.set_title(
        model_name,
        fontsize=13,
        fontweight="bold",
        pad=12
    )

    # --------------------------------------------------------
    # Grid
    # --------------------------------------------------------

    ax.grid(
        axis="x",
        linestyle=":",
        linewidth=0.6,
        alpha=0.5
    )

    ax.tick_params(
        axis="x",
        labelsize=9
    )

    ax.tick_params(
        axis="y",
        labelsize=9
    )


# ------------------------------------------------------------
# 10. Overall title
# ------------------------------------------------------------

fig.suptitle(
    "SHAP-Based Interpretation of Heart Failure Mortality "
    "Prediction Models",
    fontsize=16,
    fontweight="bold",
    y=1.02
)


# ------------------------------------------------------------
# 11. Figure note
# ------------------------------------------------------------

fig.text(
    0.5,
    -0.015,
    "Positive SHAP values indicate greater contribution toward "
    "the positive mortality outcome, whereas negative values "
    "indicate contribution toward the negative outcome. "
    "SHAP values were obtained from 5-fold out-of-fold predictions.",
    ha="center",
    fontsize=10
)


# ------------------------------------------------------------
# 12. Layout
# ------------------------------------------------------------

plt.tight_layout(
    rect=[0, 0.045, 1, 0.96]
)


# ------------------------------------------------------------
# 13. Save publication-quality files
# ------------------------------------------------------------

plt.savefig(
    "Figure_6_SHAP_Analysis_Heart_Failure.png",
    dpi=300,
    bbox_inches="tight"
)

plt.savefig(
    "Figure_6_SHAP_Analysis_Heart_Failure.pdf",
    bbox_inches="tight"
)


# ------------------------------------------------------------
# 14. Display
# ------------------------------------------------------------

plt.show()
```

```python
# ============================================================
# FIGURE 7: CALIBRATION CURVES
# HEART FAILURE DATASET
# 5-FOLD OUT-OF-FOLD PREDICTED PROBABILITIES
# Publication-ready version
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.base import clone
from sklearn.calibration import calibration_curve
from sklearn.metrics import brier_score_loss


# ------------------------------------------------------------
# 1. Five-fold stratified cross-validation
# ------------------------------------------------------------

skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)


# ------------------------------------------------------------
# 2. Generate OOF predicted probabilities
# ------------------------------------------------------------

hf_oof_probabilities = {}

for model_name, model in models.items():

    print(f"Generating OOF probabilities: {model_name}")

    model_clone = clone(model)

    # --------------------------------------------------------
    # Use predicted probability
    # --------------------------------------------------------

    y_prob_oof = cross_val_predict(
        model_clone,
        X_hf,
        y_hf,
        cv=skf,
        method="predict_proba",
        n_jobs=-1
    )[:, 1]

    hf_oof_probabilities[model_name] = y_prob_oof


# ------------------------------------------------------------
# 3. Store probabilities
# ------------------------------------------------------------

hf_oof_probabilities_df = pd.DataFrame(
    hf_oof_probabilities
)

hf_oof_probabilities_df["Actual"] = np.asarray(
    y_hf
)


# ------------------------------------------------------------
# 4. Save OOF probabilities
# ------------------------------------------------------------

hf_oof_probabilities_df.to_csv(
    "Heart_Failure_OOF_Predicted_Probabilities_5Fold.csv",
    index=False
)


# ------------------------------------------------------------
# 5. Calculate calibration and Brier scores
# ------------------------------------------------------------

calibration_results = []

n_bins = 10

for model_name, y_prob in hf_oof_probabilities.items():

    fraction_of_positives, mean_predicted_value = (
        calibration_curve(
            y_hf,
            y_prob,
            n_bins=n_bins,
            strategy="quantile"
        )
    )

    brier = brier_score_loss(
        y_hf,
        y_prob
    )

    calibration_results.append({
        "Model": model_name,
        "Brier_Score": brier
    })


calibration_results_df = pd.DataFrame(
    calibration_results
)


# ------------------------------------------------------------
# 6. Display Brier scores
# ------------------------------------------------------------

print("\nBrier Scores")
print("=" * 50)

display(
    calibration_results_df.sort_values(
        "Brier_Score"
    ).reset_index(drop=True)
)


# ------------------------------------------------------------
# 7. Plot calibration curves
# ------------------------------------------------------------

fig, ax = plt.subplots(
    figsize=(10, 8)
)


for model_name, y_prob in hf_oof_probabilities.items():

    fraction_of_positives, mean_predicted_value = (
        calibration_curve(
            y_hf,
            y_prob,
            n_bins=n_bins,
            strategy="quantile"
        )
    )

    brier = brier_score_loss(
        y_hf,
        y_prob
    )

    ax.plot(
        mean_predicted_value,
        fraction_of_positives,
        marker="o",
        linewidth=2,
        markersize=5,
        label=f"{model_name} (Brier = {brier:.3f})"
    )


# ------------------------------------------------------------
# 8. Perfect calibration reference line
# ------------------------------------------------------------

ax.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    linewidth=1.5,
    label="Perfect Calibration"
)


# ------------------------------------------------------------
# 9. Axis labels
# ------------------------------------------------------------

ax.set_xlabel(
    "Mean Predicted Probability of Death",
    fontsize=13
)

ax.set_ylabel(
    "Observed Proportion of Death Events",
    fontsize=13
)


# ------------------------------------------------------------
# 10. Title
# ------------------------------------------------------------

ax.set_title(
    "Calibration of Machine Learning Models for "
    "Heart Failure Mortality Prediction",
    fontsize=15,
    fontweight="bold",
    pad=15
)


# ------------------------------------------------------------
# 11. Axis limits
# ------------------------------------------------------------

ax.set_xlim(
    0,
    1
)

ax.set_ylim(
    0,
    1
)


# ------------------------------------------------------------
# 12. Ticks
# ------------------------------------------------------------

ax.tick_params(
    axis="both",
    labelsize=11
)


# ------------------------------------------------------------
# 13. Grid
# ------------------------------------------------------------

ax.grid(
    linestyle=":",
    linewidth=0.7,
    alpha=0.6
)


# ------------------------------------------------------------
# 14. Legend
# ------------------------------------------------------------

ax.legend(
    title="Model",
    fontsize=9,
    title_fontsize=10,
    loc="upper left",
    frameon=True
)


# ------------------------------------------------------------
# 15. Figure note
# ------------------------------------------------------------

fig.text(
    0.5,
    0.01,
    "Calibration curves are based on 5-fold stratified "
    "out-of-fold predicted probabilities. Lower Brier scores "
    "indicate better probabilistic accuracy.",
    ha="center",
    fontsize=10
)


# ------------------------------------------------------------
# 16. Layout
# ------------------------------------------------------------

plt.tight_layout(
    rect=[0, 0.045, 1, 1]
)


# ------------------------------------------------------------
# 17. Save publication-quality files
# ------------------------------------------------------------

plt.savefig(
    "Figure_7_Calibration_Curves_Heart_Failure.png",
    dpi=300,
    bbox_inches="tight"
)

plt.savefig(
    "Figure_7_Calibration_Curves_Heart_Failure.pdf",
    bbox_inches="tight"
)


# ------------------------------------------------------------
# 18. Display
# ------------------------------------------------------------

plt.show()
```
